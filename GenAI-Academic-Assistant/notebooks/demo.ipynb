```python
# Import required libraries
import sys
import os

# Add src directory to path
sys.path.append('../src')

from main import AcademicAssistant
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set up plotting style
plt.style.use('seaborn-v0_8')
sns.set_palette("husl")

print("✅ Libraries imported successfully!")
```

```python
# Initialize the Academic Assistant
# Make sure to set your OpenAI API key in environment variables
assistant = AcademicAssistant()

print("🤖 Academic Assistant initialized!")
print("Ready to demonstrate AI-powered academic assistance.")
```

## 📄 Demo 1: Text Summarization

Let's test the summarization capability with a sample academic text about machine learning.

```python
# Sample academic text for summarization
sample_text = """
Machine learning is a method of data analysis that automates analytical model building. 
It is a branch of artificial intelligence based on the idea that systems can learn from data, 
identify patterns and make decisions with minimal human intervention. Machine learning algorithms 
build a model based on training data in order to make predictions or decisions without being 
explicitly programmed to do so. Machine learning algorithms are used in a wide variety of 
applications, such as in medicine, email filtering, speech recognition, and computer vision, 
where it is difficult or unfeasible to develop conventional algorithms to perform the needed tasks.

A subset of machine learning is closely related to computational statistics, which focuses on 
making predictions using computers; but not all machine learning is statistical learning. 
The study of mathematical optimization delivers methods, theory and application domains to the 
field of machine learning. Data mining is a related field of study, focusing on exploratory 
data analysis through unsupervised learning. Some implementations of machine learning use data 
and neural networks in a way that mimics the workings of a biological brain. In its application 
across business problems, machine learning is also referred to as predictive analytics.
"""

print("📝 Original Text Length:", len(sample_text.split()), "words")
print("\n" + "="*60)
print("ORIGINAL TEXT:")
print(sample_text)
print("="*60)
```

```python
# Generate summary
summary = assistant.summarize(sample_text)

print("📄 GENERATED SUMMARY:")
print("="*40)
print(summary)
print("="*40)
print(f"\n📊 Summary Length: {len(summary.split())} words")
print(f"📊 Compression Ratio: {len(sample_text.split())/len(summary.split()):.1f}:1")
```

## ❓ Demo 2: Question Answering

Now let's test the Q&A capability using the same text as context.

```python
# Test questions for Q&A
test_questions = [
    "What is machine learning?",
    "What are some applications of machine learning?",
    "How is machine learning related to data mining?",
    "What is predictive analytics?",
    "What programming languages are best for machine learning?"  # This should return "not found"
]

print("❓ QUESTION ANSWERING DEMO")
print("="*50)

qa_results = []

for i, question in enumerate(test_questions, 1):
    answer = assistant.answer_question(sample_text, question)
    
    print(f"\n🔍 Question {i}: {question}")
    print(f"💡 Answer: {answer}")
    print("-" * 50)
    
    qa_results.append({
        'Question': question,
        'Answer': answer,
        'Answer_Length': len(answer.split())
    })

# Create DataFrame for analysis
qa_df = pd.DataFrame(qa_results)
print("\n📊 Q&A Results Summary:")
print(qa_df[['Question', 'Answer_Length']])
```

## ✍️ Demo 3: Content Generation

Let's generate structured academic content for different topics.

```python
# Topics for content generation
topics = [
    "Neural Networks",
    "Data Privacy in AI",
    "Quantum Computing"
]

print("✍️ CONTENT GENERATION DEMO")
print("="*50)

generated_content = {}

for topic in topics:
    print(f"\n🎯 Generating content for: {topic}")
    print("-" * 40)
    
    content = assistant.generate_content(topic)
    generated_content[topic] = content
    
    print(content)
    print(f"\n📊 Content Length: {len(content.split())} words")
    print("=" * 50)
```

## 📊 Performance Analysis

Let's analyze the performance of our Academic Assistant across different tasks.

```python
# Create performance analysis
performance_data = {
    'Task': ['Summarization', 'Q&A (Avg)', 'Content Generation (Avg)'],
    'Output_Length_Words': [
        len(summary.split()),
        qa_df['Answer_Length'].mean(),
        sum(len(content.split()) for content in generated_content.values()) / len(generated_content)
    ],
    'Task_Type': ['Compression', 'Extraction', 'Generation']
}

perf_df = pd.DataFrame(performance_data)

# Create visualization
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))

# Bar plot of output lengths
sns.barplot(data=perf_df, x='Task', y='Output_Length_Words', ax=ax1)
ax1.set_title('Average Output Length by Task Type')
ax1.set_ylabel('Words')
ax1.tick_params(axis='x', rotation=45)

# Pie chart of task distribution
task_counts = perf_df['Task_Type'].value_counts()
ax2.pie(task_counts.values, labels=task_counts.index, autopct='%1.1f%%')
ax2.set_title('Task Type Distribution')

plt.tight_layout()
plt.show()

print("📊 Performance Summary:")
print(perf_df)
```

## 🔍 Evaluation Metrics

Let's evaluate the quality of our outputs using basic metrics.

```python
# Basic evaluation metrics
def calculate_readability_score(text):
    """Simple readability score based on sentence and word length."""
    sentences = text.split('.')
    words = text.split()
    
    if len(sentences) == 0 or len(words) == 0:
        return 0
    
    avg_sentence_length = len(words) / len(sentences)
    avg_word_length = sum(len(word) for word in words) / len(words)
    
    # Simple readability formula (lower is better)
    readability = (avg_sentence_length * 0.39) + (avg_word_length * 11.8) - 15.59
    return max(0, readability)

# Evaluate all outputs
evaluation_results = []

# Summary evaluation
summary_score = calculate_readability_score(summary)
evaluation_results.append({
    'Output_Type': 'Summary',
    'Text': summary[:100] + '...',
    'Word_Count': len(summary.split()),
    'Readability_Score': summary_score
})

# Q&A evaluation
for i, result in enumerate(qa_results[:3]):  # First 3 Q&A pairs
    score = calculate_readability_score(result['Answer'])
    evaluation_results.append({
        'Output_Type': f'Q&A_{i+1}',
        'Text': result['Answer'][:100] + '...',
        'Word_Count': result['Answer_Length'],
        'Readability_Score': score
    })

# Content generation evaluation
for topic, content in generated_content.items():
    score = calculate_readability_score(content)
    evaluation_results.append({
        'Output_Type': f'Generated_{topic.replace(" ", "_")}',
        'Text': content[:100] + '...',
        'Word_Count': len(content.split()),
        'Readability_Score': score
    })

eval_df = pd.DataFrame(evaluation_results)

print("🔍 EVALUATION RESULTS")
print("=" * 60)
print(eval_df[['Output_Type', 'Word_Count', 'Readability_Score']])

# Visualization
plt.figure(figsize=(12, 6))
sns.scatterplot(data=eval_df, x='Word_Count', y='Readability_Score', 
                hue='Output_Type', s=100)
plt.title('Output Quality Analysis: Word Count vs Readability')
plt.xlabel('Word Count')
plt.ylabel('Readability Score (lower is better)')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```
